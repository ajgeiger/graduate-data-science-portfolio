# Prepare data for the R scripts

Saves sampled feature sets to `data/refined/` for the R classification and clustering scripts. The raw data is at https://www.kaggle.com/c/m5-forecasting-accuracy/data.

```python
# run from the project folder so data/ paths work and src/ is importable
import os
import sys

if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
sys.path.insert(0, "src")
```

```python
from m5_forecasting.features import FeaturePipeline
```

```python
pipeline = FeaturePipeline(path="data/raw/", verbose=False)
```

## Classification data

```python
classification_data = pipeline.build_classification_feature_set().dropna()
classification_data.sample(100000).to_csv("data/refined/classification_data.csv", index=False)
```

## Clustering data

The sample is small on purpose. With more rows, the R hierarchical clustering failed with:

**Error: cannot allocate vector of size 2384.2 Gb**

```python
clustering_data = pipeline.build_feature_set().dropna()
clustering_data.sample(1000).to_csv("data/refined/clustering_data.csv", index=False)
```
